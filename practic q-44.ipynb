# 26. Find the prime factors of an integer input by the user.

```python
def prime():
    num1 = input("please enter a number")
    if num1.isdigit():
        n1 = int(num1)
    
    for i in range(2,n1):
        if n1%i==0:
            return 'not a prime'
            
    return "prime"
```
