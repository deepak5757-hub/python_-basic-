```python
class calc:
    def __init__ (self,Num,user):
       self.n = Num # variable ko kuch bhi name de sakta ho but u have to call as same u store it like "n" in thi case
       self.u = user
    
    def add (self):
        Number = float(input("please enter a number that u want to add"))
        self.n+=Number
        return self.n , self.u

    def multiply(self):
        Number = float(input("please enter a number that u want to multiplication"))
        return self.n*Number , self.u
         
```

 # Duck Type 30-05-26

1) Promotes flexibility and reuseable code
2) reduces the need for complex inheritance herirarchies
3) fits python dynamic nature

```python
class school: 
    def aaluu(self):
        return "hoeee"
class lab:
    def aaluu(self):
        return "hello bhaiii"
def call_child(bacha):
    return bacha.aaluu()
```

```python
h = school()
o = lab()
```

```python
x=call_child(o)
```

```python
x
```
